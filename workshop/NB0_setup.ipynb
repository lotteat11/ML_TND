```python
import os, sys

# 1) Install packages
print('Installing packages...')
os.system('pip install -q xgboost scikit-learn pandas numpy matplotlib scipy joblib pyarrow')

# 2) Clone repo
print('Cloning workshop code...')
if not os.path.exists('/content/ML_TND'):
    os.system('git clone -q https://github.com/lotteat11/ML_TND /content/ML_TND')
else:
    os.system('cd /content/ML_TND && git pull -q')

# 3) Download data
dest = '/content/ML_TND/grace_workshop_small.parquet'
if not os.path.exists(dest):
    print('Downloading workshop data (~257 MB)...')
    ret = os.system(
        f'wget -q --show-progress -O {dest} '
        'https://github.com/lotteat11/ML_TND/releases/download/v1.0-workshop/grace_workshop_small.parquet'
    )
    if ret != 0 or not os.path.exists(dest):
        raise RuntimeError('Download failed — check your internet connection and re-run this cell.')
else:
    print('Data already downloaded.')

# 4) Verify
import importlib
missing = []
for pkg in ['xgboost', 'sklearn', 'pandas', 'numpy', 'matplotlib', 'scipy', 'joblib', 'pyarrow']:
    try: importlib.import_module(pkg)
    except ImportError: missing.append(pkg)
if missing:
    raise RuntimeError(f'Missing packages: {missing}. Re-run this cell.')
if not os.path.exists(dest):
    raise RuntimeError('Data file not found. Re-run this cell.')

print()
print('=' * 52)
print('  ✅  YOU ARE GOOD TO GO!')
print('=' * 52)
print()
print('Next: File → Open notebook → GitHub tab')
print('Paste this link to start:')
print('github.com/lotteat11/ML_TND/blob/main/workshop/NB1_baseline.ipynb')
```
